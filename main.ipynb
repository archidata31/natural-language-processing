# Natural Language Processing: Simple English Sentence Parser

### Course

Natural Language Processing (CSE 306)

### Project Objective

To build a basic Natural Language Processing system that accepts or rejects a given English sentence based on a predefined corpus of 20 simple English sentences.

The system performs:

* Tokenization
* Subject identification
* Verb identification
* Object identification
* Simple sentence parsing
* Sentence acceptance/rejection
* Reason generation for rejected sentences


```python
import re
```

# Create Your 20-Sentence Corpus

```python
corpus = [
    "Shyam eats mango.",
    "Ram plays cricket.",
    "Sita reads a book.",
    "Rohan drinks water.",
    "Priya writes a letter.",
    "Aman kicks the ball.",
    "Neha sings a song.",
    "Rahul drives a car.",
    "Anita cooks food.",
    "Vijay buys a phone.",
    "Mohan watches a movie.",
    "Riya draws a picture.",
    "Arjun opens the door.",
    "Karan cleans the room.",
    "Meena plants a tree.",
    "Raj drinks milk.",
    "Pooja makes tea.",
    "Amit flies a kite.",
    "Sneha reads a newspaper.",
    "Varun rides a bicycle."
]

print("Number of sentences in corpus:", len(corpus))

for i, sentence in enumerate(corpus, 1):
    print(i, sentence)
```

# Define the Words

```python
subjects = {
    "shyam", "ram", "sita", "rohan", "priya",
    "aman", "neha", "rahul", "anita", "vijay",
    "mohan", "riya", "arjun", "karan", "meena",
    "raj", "pooja", "amit", "sneha", "varun"
}

verbs = {
    "eats", "plays", "reads", "drinks", "writes",
    "kicks", "sings", "drives", "cooks", "buys",
    "watches", "draws", "opens", "cleans", "plants",
    "makes", "flies", "rides"
}

articles = {"a", "the"}

objects = {
    "mango", "cricket", "book", "water", "letter",
    "ball", "song", "car", "food", "phone",
    "movie", "picture", "door", "room", "tree",
    "milk", "tea", "kite", "newspaper", "bicycle"
}
```

```python
def tokenize(sentence):
    sentence = sentence.lower()
    sentence = re.sub(r'[.!?,]', '', sentence)
    tokens = sentence.split()
    return tokens
```

```python
sentence = "Sita reads a book."

print("Original sentence:", sentence)
print("Tokens:", tokenize(sentence))
```

# Identify the Role of Each Word

```python
def identify_roles(tokens):
    roles = []

    for word in tokens:

        if word in subjects:
            role = "Subject"

        elif word in verbs:
            role = "Verb"

        elif word in articles:
            role = "Article"

        elif word in objects:
            role = "Object"

        else:
            role = "Unknown"

        roles.append((word, role))

    return roles
```

```python
tokens = tokenize("Sita reads a book.")

roles = identify_roles(tokens)

for word, role in roles:
    print(word, "->", role)
```

# Simple Parser

```python
def parse_sentence(sentence):

    tokens = tokenize(sentence)

    # Check number of words
    if len(tokens) < 3:
        return False, "Sentence is too short.", None

    # Identify roles
    roles = identify_roles(tokens)

    # Check for unknown words
    for word, role in roles:
        if role == "Unknown":
            return False, f"Unknown word '{word}' is not in the corpus.", roles

    # Pattern 1: Subject + Verb + Object
    if len(tokens) == 3:
        if (
            roles[0][1] == "Subject"
            and roles[1][1] == "Verb"
            and roles[2][1] == "Object"
        ):
            return True, "Sentence follows Subject-Verb-Object structure.", roles

        else:
            return False, "Sentence does not follow Subject-Verb-Object structure.", roles

    # Pattern 2: Subject + Verb + Article + Object
    elif len(tokens) == 4:
        if (
            roles[0][1] == "Subject"
            and roles[1][1] == "Verb"
            and roles[2][1] == "Article"
            and roles[3][1] == "Object"
        ):
            return True, "Sentence follows Subject-Verb-Article-Object structure.", roles

        else:
            return False, "Sentence does not follow the expected sentence structure.", roles

    else:
        return False, "Sentence has an unsupported structure.", roles
```

# Test an Accepted Sentence

```python
sentence = "Sita reads a book."

accepted, reason, roles = parse_sentence(sentence)

print("Sentence:", sentence)
print("Accepted:", accepted)
print("Reason:", reason)

print("\nWord Roles:")
for word, role in roles:
    print(f"{word} -> {role}")
```

# Create an Interactive Program

```python
def check_sentence():

    sentence = input("Enter an English sentence: ")

    accepted, reason, roles = parse_sentence(sentence)

    print("\n--- Result ---")
    print("Sentence:", sentence)

    if accepted:
        print("Status: ACCEPTED")
        print("Reason:", reason)

        print("\nWord Roles:")
        for word, role in roles:
            print(f"{word} -> {role}")

    else:
        print("Status: REJECTED")
        print("Reason:", reason)

        if roles:
            print("\nDetected Word Roles:")
            for word, role in roles:
                print(f"{word} -> {role}")
```

```python
check_sentence()
```

# Display the Corpus Dynamically

```python
def display_corpus():
    print("\n--- Corpus ---")

    for i, sentence in enumerate(corpus, 1):
        print(f"{i}. {sentence}")
```

```python
display_corpus()
```

# Add a New Sentence

```python
def add_sentence():

    sentence = input("Enter a new sentence: ")

    accepted, reason, roles = parse_sentence(sentence)

    if accepted:
        corpus.append(sentence)
        print("Sentence added successfully!")
    else:
        print("Sentence was not added.")
        print("Reason:", reason)
```

# Main Menu

```python
def main_menu():

    while True:

        print("\n==============================")
        print("     SIMPLE NLP PARSER")
        print("==============================")

        print("1. Display Corpus")
        print("2. Check Sentence")
        print("3. Add New Sentence")
        print("4. Exit")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            display_corpus()

        elif choice == "2":
            check_sentence()

        elif choice == "3":
            add_sentence()

        elif choice == "4":
            print("Thank you for using the NLP Parser.")
            break

        else:
            print("Invalid choice. Please select 1-4.")
```

```python
main_menu()
```

3. Only predefined subjects, verbs, and objects are recognized.

4. More complex grammatical structures are not supported.

5. Synonyms and different forms of words may not be recognized unless they are added to the vocabulary.

6. The parser uses predefined rules instead of a complete natural language grammar model.

7. The system does not understand the meaning or context of a sentence.
